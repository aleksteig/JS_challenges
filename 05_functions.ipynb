## Part 1: What is a function?
You've used functions since day one: `console.log(...)`, `print(...)`, `len(...)`.
A **function** is a named piece of code that you can run (**call**) whenever you need it.
It takes **inputs** (parameters) and gives back an **output** (the return value).

```
input ──▶ [ function ] ──▶ output
"Ada" ──▶ [  greet   ] ──▶ "Hello, Ada!"
```

````python
%%node
// Define the function. This does NOT run it yet.
function greet(name) {
  return `Hello, ${name}!`;
}

// Call it, as many times as you like
console.log(greet("Ada"));
console.log(greet("Linus"));
const message = greet("Grace");
console.log(message);
````

````python
def greet(name):
    return f"Hello, {name}!"

print(greet("Ada"))
print(greet("Linus"))
message = greet("Grace")
print(message)
````

- **Define** once (`function` / `def`), **call** many times (`greet(...)`).
- `name` is a **parameter**: a variable that gets its value when the function is called.
  `"Ada"` is the **argument**: the actual value passed in.

## Part 2: Several parameters and default values

````python
%%node
function rectangleArea(width, height) {
  return width * height;
}

function greetIn(name, language = "en") {
  if (language === "no") {
    return `Hei, ${name}!`;
  }
  return `Hello, ${name}!`;
}

console.log(rectangleArea(3, 4));
console.log(greetIn("Ada"));
console.log(greetIn("Ada", "no"));
````

## Part 3: `return` vs printing (important!)

````python
%%node
// 🔮 My prediction:
function addAndPrint(a, b) {
  console.log(a + b);
}

function addAndReturn(a, b) {
  return a + b;
}

const x = addAndPrint(2, 3);
const y = addAndReturn(2, 3);
console.log("x is", x);
console.log("y is", y);
````

````python
# 🔮 My prediction:
def add_and_print(a, b):
    print(a + b)

def add_and_return(a, b):
    return a + b

x = add_and_print(2, 3)
y = add_and_return(2, 3)
print("x is", x)
print("y is", y)
````

### 🔍 Investigate
- `console.log` / `print` **shows** a value to a human. The program can't use it afterwards.
- `return` **gives the value back** to the code that called the function, so it can be stored,
  compared or passed on.
- A function without `return` gives back `undefined` (JavaScript) or `None` (Python).
- **Rule of thumb:** functions should `return` their result and let the caller decide whether to print it.

````python
%%node
// 🔮 What gets printed?
function isAdult(age) {
  return age >= 18;
  console.log("Checked the age!");
}

console.log(isAdult(20));
````

🔍 `return` ends the function **immediately**. The line after it never runs.

## Part 4: Functions calling functions

````python
%%node
function square(n) {
  return n * n;
}

function sumOfSquares(a, b) {
  return square(a) + square(b);
}

console.log(sumOfSquares(3, 4));
````

`sumOfSquares(3, 4)` pauses and calls `square(3)`, which returns 9. Then it calls `square(4)`, which
returns 16. Then it can finish and return 9 + 16 = 25.

```
sumOfSquares(3, 4)
├── square(3) → 9
├── square(4) → 16
└── returns 25
```

The computer keeps a **call stack**: a pile of functions that are waiting for an answer. You'll see the
call stack again in error messages in lesson 08.

## Part 5: Scope: where does a variable exist?
Variables created inside a function only exist inside that function. They are **local**.

````python
%%node
// 🔮 My prediction:
let message = "I am global";

function show() {
  let message = "I am local";
  console.log("inside:", message);
}

show();
console.log("outside:", message);
````

````python
%%node
// 🔮 My prediction:
function makeSecret() {
  const secret = 42;
  return secret;
}

console.log(makeSecret());
console.log(secret);
````

Now a difference between the languages. 🔮 Does this work in JavaScript? In Python?

````python
%%node
if (true) {
  let insideIf = "hi";
}
console.log(insideIf);
````

````python
if True:
    inside_if = "hi"
print(inside_if)
````

### 🔍 Investigate
- In JavaScript, `let` and `const` only exist inside the nearest `{ }`. This is called **block scope**.
  Java and C# work the same way.
- In Python, only functions create a new scope. `if` and `for` blocks don't.
- Why is scope good? You can use names like `i` or `total` in many functions without them interfering
  with each other.

## Part 6: Arrow functions

````python
%%node
// These two do the same thing
function double(n) {
  return n * 2;
}
const doubleArrow = (n) => n * 2;

console.log(double(5), doubleArrow(5));
````

Arrow functions are a short way to write small functions. You'll use them a lot in lesson 07.
Python's closest equivalent is `lambda n: n * 2`, but in Python it's normal to just use `def`.

## Part 7: Decomposition: big problems, small functions
Split a problem into small functions that each do **one thing** and have a **clear name**.
The main function then reads almost like plain English:

````python
%%node
function hasMinLength(password) {
  return password.length >= 8;
}

function hasDigit(password) {
  for (const ch of password) {
    if ("0123456789".includes(ch)) {
      return true;   // found one: no need to look further
    }
  }
  return false;
}

function isStrongPassword(password) {
  return hasMinLength(password) && hasDigit(password);
}

console.log(isStrongPassword("hello"));
console.log(isStrongPassword("hellothere"));
console.log(isStrongPassword("hello12345"));
````

### 🔍 Investigate

## 🌍 Rosetta Stone: same idea, different language
The ideas are the same in every language. Only the spelling changes. Add your own rows when you discover something new.

| Idea | JavaScript | Python |
|---|---|---|
| Define a function | `function f(a, b) { return a + b; }` | `def f(a, b):` + `return a + b` |
| Short function | `const f = (a, b) => a + b;` | `f = lambda a, b: a + b` |
| Default value | `function f(x = 1)` | `def f(x=1):` |
| No return value | `undefined` | `None` |
| Missing argument | becomes `undefined` | `TypeError` |
| What creates a new scope | every `{ }` block (`let` / `const`) | functions only |
