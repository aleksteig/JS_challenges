## Part 1: Comparisons give booleans
A decision needs a question with a yes/no answer. In code, that answer is a **boolean**: `true` or `false`.

| Meaning | JavaScript | Python |
|---|---|---|
| equal | `===` | `==` |
| not equal | `!==` | `!=` |
| greater / smaller | `>` `<` `>=` `<=` | `>` `<` `>=` `<=` |

````python
%%node
// 🔮 My prediction:
console.log(5 > 3);
console.log(5 >= 5);
console.log(5 === 5);
console.log(5 === "5");
console.log(5 == "5");
console.log("apple" < "banana");
````

````python
# 🔮 My prediction:
print(5 > 3)
print(5 == 5)
print(5 == "5")
print("apple" < "banana")
````

### 🔍 Investigate
- `===` checks that the value **and the type** are the same. `==` converts types first (just like
  `"5" + 1` in lesson 02), so `5 == "5"` is `true`. **In JavaScript, always use `===` and `!==`.**
- Python's `==` never converts types, so it behaves like JavaScript's `===`.
- Strings are compared alphabetically (more precisely: by character codes, so `"Z" < "a"`).

## Part 2: `if` / `else`
An `if` statement runs a block of code only when its condition is `true`.
- JavaScript: the condition goes in `( )`, the block in `{ }`.
- Python: a colon `:` and **indentation** (4 spaces) mark the block. `else if` is spelled `elif`.

````python
%%node
// 🔮 My prediction:
const temperature = 18;

if (temperature > 25) {
  console.log("Shorts weather ☀️");
} else if (temperature > 15) {
  console.log("Light jacket 🧥");
} else {
  console.log("Warm coat 🧣");
}
console.log("Have a nice day!");
````

````python
# 🔮 My prediction:
temperature = 18

if temperature > 25:
    print("Shorts weather ☀️")
elif temperature > 15:
    print("Light jacket 🧥")
else:
    print("Warm coat 🧣")
print("Have a nice day!")
````

### 🔍 Investigate
- Only **one** branch runs: the first one whose condition is true. The others are skipped.
- "Have a nice day!" is outside the `if`, so it always runs.
- In Python, indentation is part of the language: it decides what's inside the `if`. In JavaScript the
  `{ }` decide, and indentation is just for humans. **Indent properly anyway.**

## Part 3: and, or, not
| Meaning | JavaScript | Python |
|---|---|---|
| both must be true | `&&` | `and` |
| at least one must be true | `\|\|` | `or` |
| the opposite | `!` | `not` |

````python
%%node
// 🔮 My prediction:
const age = 20;
const hasTicket = false;
const isVip = true;

console.log(age >= 18 && hasTicket);
console.log(hasTicket || isVip);
console.log(!hasTicket);

if ((age >= 18 && hasTicket) || isVip) {
  console.log("Welcome in!");
} else {
  console.log("Sorry, you can't come in.");
}
````

````python
# 🔮 My prediction:
age = 20
has_ticket = False
is_vip = True

print(age >= 18 and has_ticket)
print(has_ticket or is_vip)
print(not has_ticket)

if (age >= 18 and has_ticket) or is_vip:
    print("Welcome in!")
else:
    print("Sorry, you can't come in.")
````

## Part 4: Truthy and falsy
An `if` doesn't need a real boolean: any value works. Values that count as `false` are called **falsy**.
Everything else is **truthy**. `Boolean(x)` / `bool(x)` shows how a value counts.
(`[]` is an empty list. You'll meet lists in lesson 06.)

````python
%%node
// 🔮 Predict true or false for each line
console.log(Boolean(0));
console.log(Boolean(42));
console.log(Boolean(""));
console.log(Boolean("hello"));
console.log(Boolean("0"));
console.log(Boolean(null));
console.log(Boolean(undefined));
console.log(Boolean([]));
````

````python
# 🔮 Predict True or False for each line
print(bool(0))
print(bool(42))
print(bool(""))
print(bool("hello"))
print(bool("0"))
print(bool(None))
print(bool([]))
````

### 🔍 Investigate
- Falsy in JavaScript: `false`, `0`, `""`, `null`, `undefined`, `NaN`.
- Falsy in Python: `False`, `0`, `""`, `None` **and empty collections** like `[]`.
- So `[]` is truthy in JavaScript but falsy in Python! A classic surprise when switching languages.

It's handy for "is there anything here?" checks:

````python
%%node
const nickname = "";

if (nickname) {
  console.log(`Hi, ${nickname}!`);
} else {
  console.log("Hi, stranger!");
}
````

## Part 5: The short `if` (ternary operator)
For choosing between two *values*:

````python
%%node
const points = 72;
const result = points >= 50 ? "pass" : "fail";
console.log(result);
````

````python
points = 72
result = "pass" if points >= 50 else "fail"
print(result)
````

Use it for short choices. For anything longer, a normal `if` / `else` is easier to read.

### 🔍 Investigate
- `lives = 0` (one `=`) **assigns** 0 to `lives` instead of comparing. 0 is falsy, so the `else` runs,
  and now the player has 0 lives but keeps playing! JavaScript gives no error. This is a **logic

- Python doesn't allow `=` in a condition, so it gives a `SyntaxError`. It protects you from this bug.
- Forgetting to indent in Python is an `IndentationError`.

## 🌍 Rosetta Stone: same idea, different language
The ideas are the same in every language. Only the spelling changes. Add your own rows when you discover something new.

| Idea | JavaScript | Python |
|---|---|---|
| Equal / not equal | `===` / `!==` | `==` / `!=` |
| And / or / not | `&&` / `\|\|` / `!` | `and` / `or` / `not` |
| If | `if (x > 0) { ... }` | `if x > 0:` + indented block |
| Else if | `else if (...) { ... }` | `elif ...:` |
| Short if | `cond ? a : b` | `a if cond else b` |
| Falsy values | `false 0 "" null undefined NaN` | `False 0 "" None` + empty collections |
| Value → boolean | `Boolean(x)` | `bool(x)` |
