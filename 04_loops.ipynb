## Part 1: `while`: repeat as long as...
Printing the numbers 1 to 5 takes five `console.log` lines. What about 1 to 1000?
A **loop** repeats a block of code.

````python
%%node
// 🔮 My prediction (what is count at the very end?):
let count = 1;

while (count <= 5) {
  console.log("count is", count);
  count = count + 1;
}
console.log("After the loop, count is", count);
````

````python
# 🔮 My prediction:
count = 1

while count <= 5:
    print("count is", count)
    count = count + 1
print("After the loop, count is", count)
````

🔍 `while` means: *as long as the condition is true, run the block again.* The condition is checked
**before** every round. When `count` becomes 6, `6 <= 5` is false and the loop stops. Each round is
called an **iteration**.

## Part 2: `for`: counting loops
Counting loops are so common that there's a shorter way to write them:

```js
for (let i = 0; i < 5; i++) { ... }
//   ─────────  ─────  ───
//   start      keep    after each
//              going   round
//              while
```

`i++` is short for `i = i + 1`. In Python you write `for i in range(5):`. `range(5)` gives
0, 1, 2, 3, 4, and `range(start, stop, step)` lets you choose more. **The stop value is not included.**

````python
%%node
// 🔮 My prediction:
for (let i = 0; i < 5; i++) {
  console.log(i);
}
````

````python
# 🔮 My prediction:
for i in range(5):
    print(i)

print("---")
for i in range(2, 10, 3):
    print(i)
````

````python
%%node
// 🔮 My prediction:
for (let i = 10; i > 0; i -= 3) {
  console.log(i);
}
````

🔍 `i -= 3` means `i = i - 3` (and `+=` means `i = i + ...`). This works in Python too.
Why start at 0? Most languages count positions from 0. Lesson 06 shows why.

## Part 4: Loop patterns
A few loop "shapes" come up again and again, in every language. Learn to recognise them.

**Accumulator**: start with an empty result and add to it in each round.

````python
%%node
// Sum of 1..100
let sum = 0;
for (let i = 1; i <= 100; i++) {
  sum += i;
}
console.log("Sum:", sum);

// Build a string
let stars = "";
for (let i = 0; i < 5; i++) {
  stars += "⭐";
}
console.log(stars);
````

**Counter**: count how many times something is true.

````python
%%node
const text = "programming is fun";
let spaces = 0;

for (const ch of text) {
  if (ch === " ") {
    spaces++;
  }
}
console.log("Spaces:", spaces);
````

🔮 **Guess first:** the same counter in Python. What will it print?

````python
text = "banana"
count_a = 0

for ch in text:
    if ch == "a":
        count_a += 1
print("Number of a's:", count_a)
````

**Search with early stop**: `break` leaves the loop immediately.

````python
%%node
// The first number whose square is bigger than 500
let n = 1;
while (true) {
  if (n * n > 500) {
    break;
  }
  n++;
}
console.log(n);
````

`continue` skips the rest of this round and goes straight to the next one:

````python
%%node
// 🔮 My prediction:
for (let i = 1; i <= 10; i++) {
  if (i % 2 === 0) {
    continue;
  }
  console.log(i);
}
````

## Part 5: Loops inside loops

````python
%%node
// 🔮 How many lines will this print?
for (let row = 1; row <= 3; row++) {
  for (let col = 1; col <= 4; col++) {
    console.log(`row ${row}, col ${col}`);
  }
}
````

🔍 The inner loop runs **completely** for each round of the outer loop: 3 × 4 = 12 lines.

## 💥 Break it: off by one

````python
%%node
// 🔮 The goal is to print exactly 5 stars. How many does it print?
for (let i = 0; i <= 5; i++) {
  console.log("⭐");
}
````

🔍 Starting at 0 *and* using `<=` gives one round too many. **Off-by-one errors** are among the most
common bugs in every language. Always check your start value and whether you need `<` or `<=`.

## 🌍 Rosetta Stone: same idea, different language
The ideas are the same in every language. Only the spelling changes. Add your own rows when you discover something new.

| Idea | JavaScript | Python |
|---|---|---|
| While loop | `while (x < 5) { ... }` | `while x < 5:` |
| Count 0 to 4 | `for (let i = 0; i < 5; i++)` | `for i in range(5):` |
| Count 1 to 10 | `for (let i = 1; i <= 10; i++)` | `for i in range(1, 11):` |
| Loop over a string / list | `for (const x of things)` | `for x in things:` |
| Add one | `i++` or `i += 1` | `i += 1` |
| Leave the loop / skip a round | `break` / `continue` | `break` / `continue` |
| Is ch in a string? | `"aeiou".includes(ch)` | `ch in "aeiou"` |
| Lower case | `s.toLowerCase()` | `s.lower()` |

```
teacher's repo on GitHub ──git pull upstream main──▶ your computer ──git push──▶ your fork on GitHub

```

A **remote** is a copy of the repository somewhere else. You have two:

```bash
git remote -v                 # list your remotes: origin (your fork) and upstream (teacher)
git fetch upstream            # download the teacher's new commits (your files don't change yet)
git log --oneline --all       # see your commits AND the teacher's
git pull upstream main        # = fetch + merge: bring the new commits into your files
```
