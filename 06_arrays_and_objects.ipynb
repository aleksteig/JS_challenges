## Part 1: Arrays (lists)
To keep track of 30 students you don't want `student1`, `student2`, ..., `student30`.
An **array** (JavaScript) or **list** (Python) holds many values **in order**.

````python
%%node
// 🔮 My prediction:
const fruits = ["apple", "banana", "cherry"];

console.log(fruits);
console.log(fruits[0]);
console.log(fruits[2]);
console.log(fruits.length);
console.log(fruits[fruits.length - 1]);
console.log(fruits[5]);
````

````python
# 🔮 My prediction:
fruits = ["apple", "banana", "cherry"]

print(fruits)
print(fruits[0])
print(fruits[-1])
print(len(fruits))
print(fruits[5])
````

### 🔍 Investigate
```
index:    0          1          2
       ┌─────────┬──────────┬──────────┐
fruits │ "apple" │ "banana" │ "cherry" │
       └─────────┴──────────┴──────────┘
```
- Positions (**indexes**) start at **0**. Think of the index as "how many steps from the start".
  So the first item is `[0]` and the last is `[length - 1]`.
- Python can count from the end: `fruits[-1]` is the last item. JavaScript has `fruits.at(-1)`.
- Asking for a position that doesn't exist: JavaScript quietly gives `undefined`, Python stops with
  an `IndexError`. Notice the pattern? JavaScript is forgiving, Python is strict.

## Part 2: Changing arrays

````python
%%node
const todo = ["learn loops", "learn functions"];

todo.push("learn arrays");          // add to the end
console.log(todo);

const removed = todo.pop();         // remove from the end
console.log(removed, todo);

todo[0] = "master loops";           // replace an item
console.log(todo);

console.log(todo.includes("learn functions"));  // is it in there?
console.log(todo.indexOf("learn functions"));   // where?

const numbers = [10, 20, 30, 40, 50];
console.log(numbers.slice(1, 3));   // from index 1 up to (not including) 3
````

## Part 3: Looping over arrays

````python
%%node
const scores = [72, 95, 88, 61];

for (const score of scores) {
  console.log("score:", score);
}

for (let i = 0; i < scores.length; i++) {
  console.log(`position ${i}: ${scores[i]}`);
}
````

````python
scores = [72, 95, 88, 61]

for score in scores:
    print("score:", score)

for i, score in enumerate(scores):
    print(f"position {i}: {score}")
````

Use `for...of` / `for x in` when you only need the items. Use the index version when you need the
positions too. Python's `enumerate` gives you both.

## Part 4: Objects (dictionaries)
Arrays are for "many things of the same kind, in order". To describe **one thing with several named
properties**, use an **object** (JavaScript) or a **dictionary** (Python): pairs of **key → value**.

````python
%%node
// 🔮 My prediction:
const student = {
  name: "Ada",
  age: 21,
  courses: ["Math", "Programming"],
};

console.log(student.name);
console.log(student["age"]);
console.log(student.courses[1]);

student.age = 22;                     // change a value
student.email = "ada@example.com";    // add a new key
console.log(student);

console.log(student.phone);
````

````python
# 🔮 My prediction:
student = {
    "name": "Ada",
    "age": 21,
    "courses": ["Math", "Programming"],
}

print(student["name"])
print(student["courses"][1])

student["age"] = 22
student["email"] = "ada@example.com"
print(student)

print(student.get("phone"))
print(student["phone"])
````

### 🔍 Investigate
- JavaScript: `student.name` or `student["name"]`. Python: `student["name"]` (the dot means something
  else in Python, see lesson 10).
- In Python the keys need quotes: `"name"`.
- A missing key: JavaScript gives `undefined`. Python's `[ ]` raises a `KeyError`, but `.get()` gives `None`.

### Looping over an object

````python
%%node
const prices = { coffee: 35, tea: 30, bun: 25 };

for (const item in prices) {
  console.log(item, "costs", prices[item]);
}

for (const [item, price] of Object.entries(prices)) {
  console.log(`${item}: ${price} kr`);
}

console.log(Object.keys(prices));
console.log(Object.values(prices));
````

````python
prices = {"coffee": 35, "tea": 30, "bun": 25}

for item in prices:
    print(item, "costs", prices[item])

for item, price in prices.items():
    print(f"{item}: {price} kr")

print(list(prices.keys()))
print(list(prices.values()))
````

> ⚠️ JavaScript trap: `for...in` loops over the **keys** of an object. `for...of` loops over the
> **items** of an array. Mixing them up is a classic bug (see 💥 below).

## Part 5: Nested data and JSON
Real data is usually an **array of objects**. This shape is so common that programs send data to each
other in exactly this format: **JSON** (JavaScript Object Notation). Almost every language can read
and write it.

````python
%%node
const students = [
  { name: "Ada", grade: 92 },
  { name: "Linus", grade: 78 },
  { name: "Grace", grade: 85 },
];

console.log(students[1].name);

for (const s of students) {
  console.log(`${s.name} got ${s.grade}`);
}

const json = JSON.stringify(students);   // data → text
console.log(json);
console.log(typeof json);

const back = JSON.parse(json);           // text → data
console.log(back[0].name);
````

````python
import json

students = [
    {"name": "Ada", "grade": 92},
    {"name": "Linus", "grade": 78},
    {"name": "Grace", "grade": 85},
]
print(students[1]["name"])

text = json.dumps(students)    # data → text
print(text)

back = json.loads(text)        # text → data
print(back[0]["name"])
````

🔍 The JSON is a **string**. A JavaScript program could send it to a Python program (over the internet,
or in a file) and the Python program could read it. That's how most apps talk to servers. You'll do it
yourself in lesson 11.

## 🌍 Rosetta Stone: same idea, different language
The ideas are the same in every language. Only the spelling changes. Add your own rows when you discover something new.

| Idea | JavaScript | Python |
|---|---|---|
| List of values | `const a = [1, 2, 3];` | `a = [1, 2, 3]` |
| First / last item | `a[0]` / `a.at(-1)` | `a[0]` / `a[-1]` |
| Length | `a.length` | `len(a)` |
| Add to end / remove from end | `a.push(x)` / `a.pop()` | `a.append(x)` / `a.pop()` |
| Is x in it? | `a.includes(x)` | `x in a` |
| Slice | `a.slice(1, 3)` | `a[1:3]` |
| Loop with index | `for (let i = 0; i < a.length; i++)` | `for i, x in enumerate(a):` |
| Key → value | `const o = { name: "Ada" };` | `d = {"name": "Ada"}` |
| Read a key | `o.name` or `o["name"]` | `d["name"]` or `d.get("name")` |
| Missing index / key | `undefined` | `IndexError` / `KeyError` |
| Loop over keys and values | `for (const [k, v] of Object.entries(o))` | `for k, v in d.items():` |
| Split text into words | `s.split(" ")` | `s.split(" ")` |
| Data → JSON text | `JSON.stringify(x)` | `json.dumps(x)` |
| JSON text → data | `JSON.parse(s)` | `json.loads(s)` |
