## Part 1: What is a program?
A **program** is a list of instructions. The computer runs them **one at a time, from top to bottom**,
and does *exactly* what they say, not what you meant.

Think of a recipe followed by a robot that takes every word literally. If the recipe says
"pour the water" before "boil the water", the robot pours cold water.

`console.log(...)` in JavaScript and `print(...)` in Python both mean: *show this on the screen.*

### 🔮 Predict
Before you run the next cell, write your prediction on the first line, after `My prediction:`.
Then run it.

````python
%%node
// 🔮 My prediction:
console.log("1 + 1");
console.log(1 + 1);
console.log("1" + "1");
````

🔮 **Guess first:** this is the same program in Python. What's different in how it's *written*?
Will it print the same thing?

````python
# 🔮 My prediction:
print("1 + 1")
print(1 + 1)
print("1" + "1")
````

### 🔍 Investigate
- Text inside quotes is a **string**. The computer doesn't calculate it. It's just characters.
- Without quotes, `1 + 1` is an **expression**: the computer works out its value, `2`.
- `"1" + "1"` joins two strings: `"11"`. The same symbol `+` does different things depending on
  *what kind of values* it gets. More about this in lesson 02.

JavaScript and Python gave the same answers, even though the code is spelled a bit differently.
**Most ideas in programming are the same in every language. Only the spelling (the *syntax*) changes.**

## Part 3: The computer remembers
Programs can store values and use them later. This is a small preview of lesson 02.

````python
%%node
// 🔮 My prediction:
let score = 0;
console.log("Score:", score);
score = score + 10;
console.log("Score:", score);
score = score + 5;
console.log("Score:", score);
````

🔍 The computer runs one line at a time and **remembers** the current value of `score`.
What's in memory at a given moment is called the program's **state**. A large part of programming
is knowing what is in memory at every line. In lesson 02 you'll learn to track it with pen and paper.

🔮 **Guess first:** the same in Python. Spot two differences in the spelling.

````python
score = 0
print("Score:", score)
score = score + 10
print("Score:", score)
score = score + 5
print("Score:", score)
````

## 💥 Part 4: Break it: reading errors
Programmers see error messages all day. An error message is not a punishment. It's the computer
telling you **what** went wrong and **where**.

### 🔮 Predict: line 3 has a typo. Will line 4 still run?

````python
%%node
// 🔮 My prediction:
console.log("Line 2 is fine");
consol.log("Line 3 has a typo");
console.log("Line 4: do I run?");
````

🔍 **Investigate.** Find these three things in the error output:
1. The **type** of error: the word ending in `Error`
2. The **message** after it: what went wrong
3. **Where**: `cell:3` means line 3 of the cell (the comment on line 1 counts as a line too)

This is a **runtime error**: it happens *while the program runs*. The program stops at that line.
Lines before it have already run, lines after it never run.

> Our `%%node` cells hide some extra lines that Node.js prints about its own internals, so the
> messages are shorter. You'll see the full version when you run `.js` files in a terminal later.

Here's the same mistake in Python. Python calls it a `NameError`. Same idea, different words:

````python
# 🔮 My prediction:
print("Line 2 is fine")
prnt("Line 3 has a typo")
print("Line 4: do I run?")
````

### 🔮 Predict: now line 3 is missing a closing quote. Will line 2 run this time?

````python
%%node
// 🔮 My prediction:
console.log("Line 2 is fine");
console.log("Line 3 is missing its closing quote);
console.log("Line 4");
````

````python
# 🔮 My prediction:
print("Line 2 is fine")
print("Line 3 is missing its closing quote)
print("Line 4")
````

🔍 Nothing ran, not even line 2! This is a **syntax error**: the code doesn't follow the grammar of the
language, so the computer can't read the program and never starts it. It's like a recipe with a
sentence that makes no sense: the robot refuses to start cooking.

| Kind of error | When does it happen? | What runs? |
|---|---|---|
| **Syntax error** | Before the program starts | Nothing |
| **Runtime error** | While the program runs | Everything before the bad line |

## 🌱 Git checkpoint
Save the notebook first (**Ctrl + S**). Git only sees saved files.

Git takes **snapshots** (called **commits**) of your files. A snapshot is made in two steps, then uploaded:

```
 your files  ──git add──▶  staging area  ──git commit──▶  history on  ──git push──▶  GitHub
(what you see)           ("put in the next              your computer              (online copy)
                           snapshot")
```

Open a terminal in the course folder (in VS Code: **Terminal → New Terminal**) and run:

```bash
git status                            # what has changed? (red = not staged yet)
git add 01_getting_started.ipynb     # put this file in the next snapshot
git status                            # now it's green: staged
git commit -m "Complete lesson 01"    # take the snapshot, with a message
git push                              # upload your snapshots to GitHub
```

Then open your repository on GitHub in the browser and find your commit.

> 💡 **Habit:** run `git status` before and after every Git command. It always tells you where you are
> and often what to do next.
