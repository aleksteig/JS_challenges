## Part 1: Values and types
Every value has a **type**. The type decides what you can do with the value: you can multiply numbers,
but you can't multiply names.

| JavaScript type | Example | Python type | Example |
|---|---|---|---|
| `number` | `42`, `3.14` | `int` / `float` | `42` / `3.14` |
| `string` | `"hello"` | `str` | `"hello"` |
| `boolean` | `true`, `false` | `bool` | `True`, `False` |
| `undefined`, `null` | `undefined`, `null` | `NoneType` | `None` |

### 🔮 Predict
`typeof` (JavaScript) and `type()` (Python) tell you the type of a value.

````python
%%node
// 🔮 My prediction:
console.log(typeof 42);
console.log(typeof 3.14);
console.log(typeof "42");
console.log(typeof true);
console.log(typeof undefined);
console.log(typeof null);
````

````python
# 🔮 My prediction:
print(type(42))
print(type(3.14))
print(type("42"))
print(type(True))
print(type(None))
````

### 🔍 Investigate
- JavaScript has **one** type for all numbers. Python has two: `int` for whole numbers and `float` for
  decimals. Java and C# separate them too.
- `"42"` with quotes is a **string**, not a number.
- `typeof null` says `"object"`. That's a mistake from the first version of JavaScript in 1995. It was
  never fixed because fixing it would break old websites. Just remember it.

## Part 2: Variables
A **variable** is a name for a value, so you can use the value later. Think of a label stuck on a box.

- JavaScript: `let` for a variable you will change, `const` for one you won't. Prefer `const` and
  use `let` only when you need to change it. (Old code uses `var`. Don't.)
- Python: no keyword, just `name = value`.
- Naming style: JavaScript uses `camelCase` (`firstName`), Python uses `snake_case` (`first_name`).

````python
%%node
const name = "Ada";
let age = 36;
console.log(name, "is", age, "years old");

age = age + 1;
console.log("Next year", name, "will be", age);
````

````python
name = "Ada"
age = 36
print(name, "is", age, "years old")

age = age + 1
print("Next year", name, "will be", age)
````

### `=` does not mean "equals"
In maths, `x = x + 1` is impossible. In programming, `=` means **assign**: *first work out the right
side, then store the result under the name on the left.* Read `age = age + 1` as
"age becomes age plus one".

## Part 4: Operators

````python
%%node
// 🔮 My prediction:
console.log(7 + 2);
console.log(7 - 2);
console.log(7 * 2);
console.log(7 / 2);
console.log(7 % 2);   // % is "remainder"
console.log(2 ** 3);  // ** is "to the power of"
````

````python
# 🔮 My prediction:
print(7 / 2)
print(7 // 2)   # what could // be?
print(7 % 2)
print(2 ** 3)
````

### 🔍 Investigate
- `%` gives the **remainder** after division: 7 = 3 · 2 + **1**. It's surprisingly useful:
  `n % 2` is `0` exactly when `n` is even.
- Python has `//` for whole-number division. JavaScript doesn't; use `Math.floor(7 / 2)`.

## Part 5: Strings
Join strings with `+`, or put values *inside* a string with a **template literal** (JavaScript, uses
backticks `` ` ``) or an **f-string** (Python).

````python
%%node
const firstName = "Grace";
const lastName = "Hopper";
const year = 1906;

console.log(firstName + " " + lastName);
console.log(`${firstName} ${lastName} was born in ${year}.`);
console.log(`In 2000 she would have turned ${2000 - year}.`);
console.log(firstName.length);
console.log(firstName.toUpperCase());
````

````python
first_name = "Grace"
last_name = "Hopper"
year = 1906

print(first_name + " " + last_name)
print(f"{first_name} {last_name} was born in {year}.")
print(f"In 2000 she would have turned {2000 - year}.")
print(len(first_name))
print(first_name.upper())
````

## Part 6: Mixing types: a big difference between languages

````python
%%node
// 🔮 My prediction:
console.log("5" + 1);
console.log("5" - 1);
console.log("5" * 2);
console.log(Number("5") + 1);
console.log(String(5) + 1);
````

````python
# 🔮 My prediction:
print(int("5") + 1)
print(str(5) + "1")
print("5" + 1)
````

### 🔍 Investigate
- JavaScript tries to be helpful and quietly **converts** values (this is called *type coercion*).
  `"5" + 1` becomes `"51"` because `+` with a string means "join". `"5" - 1` becomes `4` because `-`
  only works with numbers. This "help" causes a lot of bugs.
- Python refuses and raises a `TypeError`. You must convert yourself with `int()`, `float()` or `str()`.
- **A good habit in every language: convert types yourself.** In JavaScript: `Number("5")`, `String(5)`.

## Part 7: Decimals aren't exact

````python
%%node
// 🔮 My prediction:
console.log(0.1 + 0.2);
````

````python
# 🔮 My prediction:
print(0.1 + 0.2)
````

🔍 This is not a bug in JavaScript or Python. Computers store decimals in binary, and 0.1 can't be
stored exactly in binary, just like 1/3 can't be written exactly as a decimal (0.3333…). Almost every
language does this. Programs that handle money often count in whole cents instead.

## 🌍 Rosetta Stone: same idea, different language
The ideas are the same in every language. Only the spelling changes. Add your own rows when you discover something new.

| Idea | JavaScript | Python |
|---|---|---|
| Create a variable | `let x = 5;` | `x = 5` |
| Constant | `const PI = 3.14;` | `PI = 3.14` (by convention only) |
| Numbers | `number` (one type) | `int` and `float` |
| True / false | `true` / `false` | `True` / `False` |
| Nothing | `undefined`, `null` | `None` |
| Type of a value | `typeof x` | `type(x)` |
| Values inside text | `` `Hi ${name}` `` | `f"Hi {name}"` |
| Length of a string | `s.length` | `len(s)` |
| Whole-number division | `Math.floor(7 / 2)` | `7 // 2` |
| Remainder | `7 % 2` | `7 % 2` |
| Text → number | `Number("5")` | `int("5")`, `float("5")` |
| Number → text | `String(5)` | `str(5)` |
| `"5" + 1` | `"51"` (converts quietly) | `TypeError` |
| Naming style | `camelCase` | `snake_case` |
