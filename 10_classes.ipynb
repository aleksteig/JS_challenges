## Part 2: Classes
A **class** is a blueprint. The objects you make from it are called **instances**.
Think of a cookie cutter (the class) and the cookies (the instances).

````python
%%node
class BankAccount {
  constructor(owner, balance = 0) {
    this.owner = owner;
    this.balance = balance;
  }

  deposit(amount) {
    if (amount <= 0) {
      throw new Error("Deposit must be positive");
    }
    this.balance += amount;
  }

  withdraw(amount) {
    if (amount > this.balance) {
      throw new Error("Not enough money");
    }
    this.balance -= amount;
  }

  describe() {
    return `${this.owner} has ${this.balance} kr`;
  }
}

const ada = new BankAccount("Ada", 100);
const linus = new BankAccount("Linus");

ada.deposit(50);
linus.deposit(20);
console.log(ada.describe());
console.log(linus.describe());

try {
  linus.withdraw(1000);
} catch (err) {
  console.log("⚠️", err.message);
}
````

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            raise ValueError("Not enough money")
        self.balance -= amount

    def describe(self):
        return f"{self.owner} has {self.balance} kr"


ada = BankAccount("Ada", 100)
linus = BankAccount("Linus")

ada.deposit(50)
linus.deposit(20)
print(ada.describe())
print(linus.describe())

try:
    linus.withdraw(1000)
except ValueError as err:
    print("⚠️", err)
````

### 🔍 Investigate

| | JavaScript | Python |
|---|---|---|
| the setup method | `constructor(...)` | `__init__(self, ...)` |
| "the current object" | `this` (automatic) | `self` (you write it as the first parameter) |
| create an instance | `new BankAccount("Ada")` | `BankAccount("Ada")` |

- Each instance has its **own** data: `ada.balance` and `linus.balance` are separate.
- The methods protect the data: `withdraw` refuses to go below zero. Code outside the class doesn't need
  to know *how*. It just calls `withdraw`. This is called **encapsulation**.

## Part 4: Inheritance
A class can **extend** another class. It gets everything from the parent and can add or change things.

````python
%%node
class Animal {
  constructor(name) {
    this.name = name;
  }
  speak() {
    return `${this.name} makes a sound`;
  }
}

class Dog extends Animal {
  speak() {
    return `${this.name} says Woof!`;
  }
}

class Cat extends Animal {
  constructor(name, indoor) {
    super(name);           // run the parent's constructor first
    this.indoor = indoor;
  }
  speak() {
    return `${this.name} says Meow${this.indoor ? " (from the sofa)" : ""}`;
  }
}

const animals = [new Animal("Generic"), new Dog("Rex"), new Cat("Misty", true)];
for (const animal of animals) {
  console.log(animal.speak());
}
````

````python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return f"{self.name} makes a sound"


class Dog(Animal):
    def speak(self):
        return f"{self.name} says Woof!"


class Cat(Animal):
    def __init__(self, name, indoor):
        super().__init__(name)
        self.indoor = indoor

    def speak(self):
        return f"{self.name} says Meow{' (from the sofa)' if self.indoor else ''}"


for animal in [Animal("Generic"), Dog("Rex"), Cat("Misty", True)]:
    print(animal.speak())
````

🔍 The loop calls `speak()` without knowing which kind of animal it has, and each object answers in its
own way. This is called **polymorphism**. Use inheritance sparingly: only when "a Dog **is an** Animal"
really is true.

## Part 6: When should you use a class?
Use a class when you have **many objects of the same kind** that have **both data and behaviour**
(bank accounts, players in a game, shapes in a drawing app). For simple data, a plain object or
dictionary is enough. For a calculation, a plain function is enough. Don't turn everything into a class.

### 1. Counter

````python
%%node
class Counter {
  // ✏️ constructor(start): value starts at `start`, or 0 if no start is given
  // ✏️ increment(): value goes up by 1
  // ✏️ decrement(): value goes down by 1, but never below 0
  // ✏️ reset(): value goes back to 0
}

const c = new Counter();
c.increment();
c.increment();
c.decrement();
check(c.value, 1);
c.decrement();
c.decrement();
check(c.value, 0);      // never below 0

const d = new Counter(10);
d.increment();
check(d.value, 11);
d.reset();
check(d.value, 0);
````

````python
%%node
class Shape {
  area() {
    return 0;
  }
  describe() {
    return `${this.constructor.name} with area ${this.area()}`;
  }
}

class Rectangle extends Shape {
  // ✏️ constructor(width, height) and area()
}

class Square extends Rectangle {
  // ✏️ constructor(side): a square is a rectangle with equal sides. Use super(side, side)
}

class Circle extends Shape {
  // ✏️ constructor(radius) and area(): Math.PI * r * r, rounded to 2 decimals
  //    Hint: Math.round(x * 100) / 100
}

check(new Rectangle(3, 4).area(), 12);
check(new Square(5).area(), 25);
check(new Circle(1).area(), 3.14);
check(new Square(2).describe(), "Square with area 4");
````

````python
# 🔮 My prediction:
class Player:
    def __init__(self, name):
        self.name = name
        self.score = 0

    def add_points(points):
        self.score += points


p = Player("Ada")
p.add_points(10)
````

### 🔍 Investigate
- Forgetting `new` in JavaScript: a class can only be used with `new`.
- Forgetting `this.`: `score` alone is a variable that doesn't exist. The score lives on the object:
  `this.score`.
- Forgetting `self` in Python: Python always passes the object as the first argument, so
  `p.add_points(10)` really sends **two** arguments, `p` and `10`.

## 🌍 Rosetta Stone: same idea, different language
The ideas are the same in every language. Only the spelling changes. Add your own rows when you discover something new.

| Idea | JavaScript | Python |
|---|---|---|
| Define a class | `class Dog { }` | `class Dog:` |
| Constructor | `constructor(name) { this.name = name; }` | `def __init__(self, name): self.name = name` |
| The current object | `this` | `self` |
| Method | `speak() { ... }` | `def speak(self): ...` |
| Create an instance | `new Dog("Rex")` | `Dog("Rex")` |
| Inherit | `class Dog extends Animal` | `class Dog(Animal):` |
| Call the parent's constructor | `super(name)` | `super().__init__(name)` |
