A real program can ask the user questions. This can't run inside a notebook cell, so save it as
`quiz.mjs` and run it in a terminal with `node quiz.mjs`:

```js
// quiz.mjs
import * as readline from "node:readline/promises";

const rl = readline.createInterface({ input: process.stdin, output: process.stdout });
const answer = await rl.question("What is 2 + 2? ");
console.log(answer.trim() === "4" ? "Correct! 🎉" : "Not quite.");
rl.close();
```

> Files ending in `.mjs` can use `import`. In a project with a `package.json` that contains
> `"type": "module"`, normal `.js` files can too.

The same in Python (`python quiz.py`):


## Part 3: The professional workflow
```
Issue #1 "Load questions from a JSON file"
   └─▶ branch 1-load-questions
         └─▶ commits
               └─▶ pull request "Load questions from JSON. Closes #1"
                     └─▶ review ─▶ merge ─▶ issue #1 closes automatically
```

1. Create a **new repository** on GitHub, with a README and a `.gitignore` (choose the *Node* template).
2. Write your plan as **Issues** (the *Issues* tab): one issue per feature, each small enough to finish
   in one session.
3. For every issue: branch → commits → pull request with `Closes #<number>` in the description → review → merge.
4. When you're done, create a **release** on GitHub (tag `v1.0`).
5. A web project? **Settings → Pages → Deploy from branch `main`** gives you a public URL.

```bash
git clone https://github.com/<you>/<project>.git
```
